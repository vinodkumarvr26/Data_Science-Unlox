```python
n= int(input("Enter value: "))
def is_prime(num):
    if num<2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True
print("Prime" if is_prime(n) else "Not Prime")
prime_sum = 0
for k in range(2, n+1):
    if is_prime(k):
        prime_sum = prime_sum + k
print("Sum =", prime_sum)
```

Q45. Caesar Cipher Encryptor. Take a message and a shift value N as input. Encrypt by shifting each 
letter forward by N positions in the alphabet (wrapping around, Z + 1 = A). Preserve case. Keep digits, 
spaces, and special characters UNCHANGED. Example: "Hello, India!" with shift 3 → "Khoor, Lqgld!" 

```python
message = input("Enter message: ")
shift = int(input("Enter shift value: "))
encrypted = ""
for ch in message:
    if 'A' <= ch <= 'Z':   
        encrypted = encrypted+chr((ord(ch)-ord('A')+shift)%26+ord('A'))
    elif 'a' <= ch <= 'z':
        encrypted= encrypted+ chr((ord(ch)-ord('a')+shift)%26+ord('a'))
    else:
        encrypted += ch
print("Encrypted message:", encrypted)
```

Q46. Write a function `calculate_gst(amount, gst_rate=18, gst_inclusive=False)`: - If gst_inclusive=False: amount is base, add GST on top -
If gst_inclusive=True: amount already includes GST, extract it - Return a tuple: (base_amount, gst_amount, total_amount) - All values rounded
to 2 decimals 
Test cases: 
calculate_gst(1000) → (1000.00, 180.00, 1180.00) 
calculate_gst(1180, gst_inclusive=True) → (1000.00, 180.00, 1180.00) 
calculate_gst(500, gst_rate=5)   → (500.00, 25.00, 525.00)

```python
def calculate_gst(amount, gst_rate=18, gst_inclusive=False):
    if not gst_inclusive:
        base = amount
        gst = round(base*gst_rate/100, 2)
        total = round(base+gst, 2)
    else:
        total = amount
        base = round(total/(1+gst_rate/100), 2)
        gst = round(total-base, 2)
    return (round(base, 2), round(gst, 2), round(total, 2))

print(calculate_gst(1000))                    
print(calculate_gst(1180, gst_inclusive=True)) 
print(calculate_gst(500, gst_rate=5))        
```

Q47. Write a RECURSIVE function `power(base, exp)` that computes base raised to exp, WITHOUT 
using the `**` operator and WITHOUT using `pow()`. Handle exp = 0 (return 1) and negative exp 
(return 1/power(base, -exp)). Test with: power(2, 10), power(5, 0), power(3, -2).

```python
def power(base, exp):
    if exp==0:
        return 1
    elif exp<0:
        return 1/power(base, -exp)
    else:
        return base*power(base, exp - 1)
print(power(2, 10))   
print(power(5, 0))  
print(power(3, -2)) 
```
